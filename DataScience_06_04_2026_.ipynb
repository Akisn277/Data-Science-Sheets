### Dataset Loading

```python
import pandas as pd

url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df = pd.read_csv(url)

# Keep only useful columns
df = df[["country", "year", "co2", "gdp", "population"]]

# Drop missing values
df = df.dropna()

# Optional: reduce size (faster plotting)
df = df.sample(1000)

df.head()
```

### Scatterplot

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x="gdp", y="co2", data=df)
plt.title("GDP vs CO2 Emissions")
plt.show()
```

### Lineplot

```python
sns.lineplot(x="year", y="co2", data=df)
plt.title("CO2 Emissions Over Time")
plt.show()
```

### Histogram

```python
sns.histplot(df["co2"], kde=True)
plt.title("Distribution of CO2 Emissions")
plt.show()
```

### Boxplot

```python
sns.boxplot(y="co2", data=df)
plt.title("Boxplot of CO2 Emissions")
plt.show()
```

### Pair Plot

```python
sns.pairplot(df[["co2", "gdp", "population"]])
plt.show()
```

### 3D Scatterplot

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(df["gdp"], df["population"], df["co2"])

ax.set_xlabel("GDP")
ax.set_ylabel("Population")
ax.set_zlabel("CO2")

plt.title("3D Scatter Plot")
plt.show()
```

### 3D Lineplot

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot(df["gdp"], df["population"], df["co2"])

ax.set_xlabel("GDP")
ax.set_ylabel("Population")
ax.set_zlabel("CO2")

plt.title("3D Line Plot")
plt.show()
```

### 3D Surface Plot

```python
import numpy as np
from scipy.interpolate import griddata

# Take smaller sample for surface
sample = df.sample(500)

X = sample["gdp"]
Y = sample["population"]
Z = sample["co2"]

# Create grid
xi = np.linspace(X.min(), X.max(), 50)
yi = np.linspace(Y.min(), Y.max(), 50)
xi, yi = np.meshgrid(xi, yi)

# Interpolate Z values
zi = griddata((X, Y), Z, (xi, yi), method='linear')

# Plot surface
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(xi, yi, zi)

ax.set_xlabel("GDP")
ax.set_ylabel("Population")
ax.set_zlabel("CO2")

plt.title("3D Surface Plot")
plt.show()
```
